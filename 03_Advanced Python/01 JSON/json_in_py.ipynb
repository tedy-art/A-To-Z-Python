# Python JSON(JavaScript Object Notation)
- JSON is a syntax for storing and exchanging data.
- JSON is text, written with JS object notation

## JSON in Python
- Python has a in-built package called `json`, which can be used to work with JSON data.
```python
import json
```

## Parse JSON - Convert from JSON to Python:
- If you have a JSON string, you can parse it by using the `json.loads()` method.

````python
# Convert from JSON to Python
import json

# Some JSON
x = '{"name": "John", "age": 30, "city": "New York"}'

# parse x
y = json.loads(x)

# the result is a python dictionary
print(y["age"])
````

## Key Functions:
1. **json.dumps():** Python object ko JSON string mein convert karta hai.
2. **json.loads():** JSON string ko Python object mein convert karta hai.
3. **json.dump():** Python object ko JSON format mein file mein likhta hai.
4. **json.load():** File se JSON data ko read karke Python object mein convert karta hai.

## Convert Python to JSON
- If you have a python object, you can convert it into a JSON string by using the `json.dumps()` method.

````python
import json

# python Dictionary
x = {
    "name": "John",
    "age": 36,
    "city":"New York"
}

# convert into JSON
y = json.dumps(x)
print(y)
````

Note:
---
We can convert py objects of the following types, into JSON strings:
- dictionary
- list
- tuple
- string
- int
- float
- True
- False
- None

````python
import json

print(json.dumps({"name":"john","age":36}))
print(json.dumps(["apple", "banana"]))
print(json.dumps(("apple", "banana")))
print(json.dumps("Hello"))
print(json.dumps(42))
print(json.dumps(31.76))
print(json.dumps(True))
print(json.dumps(False))
print(json.dumps(None))
````

### Convert a Python object containing all the legal data types:

````python
import json

# Py dictionary
x = {
    "name":"John",
    "age":36,
    "married": True,
    "divorced": False,
    "Children": ("Anne", "Billy"),
    "pets": None,
    "cars": [
        {"model":"BMW 230", "mpg": 27.5},
        {"model":"Ford Edge", "mpg": 24.1}
    ]
}

# Convet Dict to JSON
print(json.dumps(x))
````

## Format the result

````python
# use the `indent` parameter to define the number of indents:
print(json.dumps(x, indent=4))
````

- You can also define the separators, default value is (", ", ": "), which means using a comma and a space to separate each object, and a colon and a space to separate keys from values:

````python
print(json.dumps(x, indent=4, separators=(".", "=")))
````

## Order the Result
- The json.dumps() method has parameters to order the keys in the result:

- Example
- Use the `sort_keys` parameter to specify if the result should be sorted or not:

````python
print(json.dumps(x, indent=4, sort_keys=True))
````

## json.dump()
Writes a Python object to a `file` in JSON format.

````python
import json

# Py dictionary
data = {
    "name":"John",
    "age":36,
    "married": True,
    "divorced": False,
    "Children": ("Anne", "Billy"),
    "pets": None,
    "cars": [
        {"model":"BMW 230", "mpg": 27.5},
        {"model":"Ford Edge", "mpg": 24.1}
    ]
}

with open("data.json", "w") as file:
    json.dump(data, file)
````

# most use methods in json py

````python
import json

# Example Python object (dictionary)
data = {
    "name": "Alice",
    "age": 25,
    "city": "New York"
}

# Convert Python object to JSON string
json_str = json.dumps(data)
print("JSON String:", json_str)

# Write JSON string to a file
with open("data.json", "w") as file:
    json.dump(data, file)

# Read JSON data from a file
with open("data.json", "r") as file:
    loaded_data = json.load(file)
    print("Loaded Data:", loaded_data)

# Convert JSON string back to Python object
python_obj = json.loads(json_str)
print("Python Object:", python_obj)
````
