# Math

Python has a set of built-in math functions, including an extensive math module, that allow you to perfrom mathematical tasks on numbers.

## Built in Math functions:
- The `min()`and `max()` functions can be used to find the lowest or highest value in an iterable.

````python
x = min(5, 62, 4)
y = max(5, 62, 4)
print(f"min element is {x}")
print(f"max element is {y}")
````

The `abs()` func returns the abosulute(+ve) value of the specified numbers:

````python
a = abs(-7.45)

print(a)
````

The `pow(x, y)` function returns the value of x to the power of `y(xy)`.

````python
print(pow(3, 4))
````

## The Math Module
- Python has also a built-in module called `math`, Which extends the list of mathematical functions.
- To use it, you must import the `math` modules:

```python
import math
```

- When you have imported the math module, you can start using methods and constrants of the module. 

1. math.sqrt()
   - return square root of a number.

````python
import math

x = math.sqrt(64)
print(x)
````

2. math.ceil():
    - method number ko upar ki taraf round karta hai uske nearest integer tak, aur result return karta hai.

````python
import math

x = math.ceil(1.4)
print(x)
````

3. math.floor():
    - method number ko neeche ki taraf round karta hai uske nearest integer tak, aur result return karta hai.

````python
import math

x = math.floor(1.4)
print(x)
````

4. math.pi:
   - return 3.14

````python
import math

x = math.pi
print(x)
````
