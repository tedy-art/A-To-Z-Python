# Python Modules

## What is Module?

- Module in python is simply a file containing python definations and statements.
- Modules allows you to organize your code into manageable sections.
- They help in code reuse and improve maintainbility.

### Key Points :
- **importing Modules :** You can import module using the `import` statement.
```python
    import mudule_name
```
- **Built-in modules :** Python comes with larger standard library of built-in modules like `math`, `os`, `sys` etc
- **Custom Modules :** You can create your own modules by saving a python file with function, variables and classes.

## Create a module
- To create a module just save the code you want in a file with the file extention `.py`.

````python
# save file with name `mymodule.py`
def greeting(name):
    return f"Hello, {name}"
````

## Import and use it in another script:
- Now we can use the module we just created, by using the `import` statement.

````python
import mymodule

mymodule.greeting("Tejas")
````

**Note :**
---
When using a function from modules, use the syntax:
```python
module_name.function_name
```

## Variables in modules
- modules can contions varibales of all types (arrays, dictionaries, objects etc)

````python
# Save this code in the file mymodule1.py

person1 = {
  "name": "John",
  "age": 36,
  "country": "Norway"
}
````

````python
import mymodule1

a = mymodule1.person["age"]
print(a)
````

### Naming a Module
- you can name the module file whatever you like, but it must have the file extension `.py`.

### Re-naming a Module
- You can create an `alias` when you import a module, by using the `as` keyword:

````python
import mymodule1 as mx

a = mx.person['age']
print(a)
````

### Built-in modules
- There are several built-in modules in Python, which you can import whatever you like.

````python
import platform

x = platform.system()
print(x)
````

### Using the `dir()` functions:


````python
import platform

x = dir(platform)
print(x)
````

### Import from Module
- you can choose to import only parts from a modules, by using the `from` keyword

````python
# save file as "mymodule2.py"

def greeting(name):
    print(f"Hello, {name}")

person = {
    "name": "John",
    "age": 36,
    "country": "Norway"
}
````

````python
from mymodule2 import person

print(person["age"])
````

### Benefits of Using Modules:

**Code Reusability:** Write once, use anywhere.

**Namespace Management:** Prevents naming conflicts.

**Organization:** Helps in organizing the code better.
