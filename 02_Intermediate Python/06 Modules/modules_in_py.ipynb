# Python Modules

## What is Module?

- Module in python is simply a file containing python definations and statements.
- Modules allows you to organize your code into manageable sections.
- They help in code reuse and improve maintainbility.

### Key Points :
- **importing Modules :** You can import module using the `import` statement.
```python
    import mudule_name
```
- **Built-in modules :** Python comes with larger standard library of built-in modules like `math`, `os`, `sys` etc
- **Custom Modules :** You can create your own modules by saving a python file with function, variables and classes.

## Create a module
- To create a module just save the code you want in a file with the file extention `.py`.

````python
# save file with name `mymodule.py`
def greeting(name):
    return f"Hello, {name}"
````
