# Python Scope

A variable is only available from the region it is created. This is called `scope`.

In python, there are four types of scopes avialable. These scopes determine the visibility and lifetime of variables:

1. Local Scope:
    - variable declared inside a function.
    - accessible only within that function

```python
def myfunc():
    # variable created inside function belong to the local scope of that function, and can only be used inside that function.
    x = 300
    print(x)

myfunc()
```

2. Enclosing(Nonlocal) Scope:
    - Variables in the local scope of enclosing function(nested functions).
    - Accessible within neasted function.

```python
def outer_function():
    x = "outer"
    # Nested finction ka ek enclosing scope hota hai, jo outer function ka scope hota hai.
    def inner_function():
        # Nonlocal keyword aapko nested functions ke saath kaam karne ko allow kar deta hai.
        nonlocal x
        x = "inner"
    inner_function()
    print(x)

outer_function()
```

3. Global Scope:
   - Variables declared at top level of a module or script.
   - Accessible throughout the module/script.

```python
x = 300

def myfunc():
    print(x)

def yourfunc():
    print(x)

myfunc()
yourfunc()
print(x)
```

3. Global Keyword:
    - If you need to create a global variable, but are stuck in the local scope, you can use the `global` keyword.

```python
def mufunc():
    global x
    x = 300
    print(x)

myfunc()
print(x) # priting x globally
```

4. Built-in Scope:
   - Contains python's built-in functions and exceptions.
   - Always avialable in any scope.

```python
print(len)
```
