# Dates in Python

- A date in python is not a data type of it's own, but we can import a module named `datetime` to work with dates as date objects.

````python
#  import the datetime module and display the current date:
import datetime

x = datetime.datetime.now()
print(x)
````

````python
today = datetime.date.today()

print(today)
````

### Creating Date Objects
- To create date object, we use the `datetime()` class (constructor) of the `datetime` modules.
- The `datetime()` class requires three parameters to create a date: year, month, day

````python
import datetime

x = datetime.datetime(2020, 5, 18)
print(x)
````

The `datetime()` class also takes parameter for time and timezone (hour, minute, seconds, microseconds, tzone), but they are optional, and has a default value of `0`,(`None` for timezone).

## Key classes and functions

1. datetime.date
   - represents a date(year, month, day)

````python
from datetime import date
today = date.today()
print(today)
````

2. datetime.time
   - represents a time(hour, minute, second, microsecond)

````python
from datetime import time

now = time(12, 30, 45)
print(now)
````

3. datetime.datetime
    - Combines date and time

````python
from datetime import datetime

now = datetime.now()
print(now)
````

4. datetime.timedelta
   - Represents the difference between two dates or times.

````python
from datetime import timedelta
delta = timedelta(days=5)
print(delta)
````

5. Formatting dates and times
   - using `strftime` to format dates and times

````python
from datetime import datetime

now = datetime.now()
formetted = now.strftime("%Y-%m-%d %H:%M:%S")
print(formetted)
````

6. Parsing dates and times
   - Using `strptime` to parse string into datetime objects.

````python
from datetime import datetime

date_str = "2024-06-26 15:30:00"
parsed_date = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")
print(parsed_date)
````

## The strftime() method

- The `datetime` objects has a method for formatting date objects into readable string.
- The method is called `strftime()`, and takes one parameter, `format`, to specify of the returned string:

````python
import datetime

x = datetime.datetime(2018, 6, 1)
print(x.strftime("%B"))
````

## The strptime() method
- `strptime` ek function hai jo `datetime` module main hota hai. Yeh function ek string ko `datetime` object main convert karta hai based on a `specified format`.

- Syntax:
```python
datetime.strptime(date_string, format)
```
- **date_string:** Date string jo aap convert karna chahte ho.

- **format:** String ka format jo aapko date_string main diya gaya hai.

````python
from datetime import datetime

date_str = "2024-06-26 15:30:00"
date_obj = datetime.strptime(date_str, "%Y-%m-%d %H:%M:%S")
print(date_obj)  # Outputs: 2024-06-26 15:30:00
````

````python
date_str = "Wed, 26 Jun 2024 15:30:00"
date_obj = datetime.strptime(date_str, "%a, %d %b %Y %H:%M:%S")
print(date_obj)  # Outputs: 2024-06-26 15:30:00
````
