# 1. Find the Second Largest Element in a List

# Brute Force

```python
def find_second_largest(nums):
    if len(nums) < 2:
        return None
    first = second = float('-inf')
    for i in range(len(nums)):
        if nums[i] > first:
            second = first
            first = nums[i]
        elif nums[i] > second and nums[i] != first:
            second = nums[i]
            
    return second if second != float('-inf') else None
    
        
numbers = [1, 3, 4, 5, 0, 2]
print(find_second_largest(numbers))
```

# Optimized Solutions

```python
def find_second_largest_optimized(nums):
    if len(nums) < 2:
        return None
    
    first = second = float('-inf')
    for num in nums:
        if num > first:
            second = first
            first = num
        elif num > second and num != first:
            second = num
    
    return second if second != float('-inf') else None

# Example usage:
numbers = [1, 3, 4, 5, 0, 2]
result2 = find_second_largest_optimized(numbers)
print(result2)
```
